```python
import pandas as pd

# Read with a encoding
df = pd.read_csv(
    "C:/Users/laneg/Downloads/ecommerce-sales-analysis-dashboard/data/superstore_dataset_raw.csv",
    encoding="latin1"
)

# Remove non-breaking spaces explicitly
df = df.replace(u'\xa0', ' ', regex=True)

# Save clean UTF-8 file
df.to_csv(
    "C:/Users/laneg/Downloads/ecommerce-sales-analysis-dashboard/data/superstore_dataset_cleaned.csv",
    index=False,
    encoding="utf-8"
)
```
